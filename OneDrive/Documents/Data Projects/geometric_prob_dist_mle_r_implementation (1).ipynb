## MLE of a Geometric Probability Distribution
#### Parameter Estimation

### Key terms:
**Maximum Likelihood Estimator (MLE):**

* This is a method for estimating the parameters of a statistical model.

* The basic idea behind maximum likelihood estimation is to find the values of the model parameters that maximize the likelihood function, which measures how well the model explains the observed data.

**Likelihood Function**:

* Given a statistical model with parameters (denoted by θ), the L(θ) shows how well the model explains the observed data.

* It represents the probability of observing the given data under different values of the parameters

**Maximum Likelihood Estimation**:

* The goal is to find the values of the parameters (^θ) that maximize the likelihood function.

**Log-Likelihood Function**:

* Due to mathematical convenience, the log-likelihood function (ℓ(θ)) is often used instead of the likelihood function.

* Taking the logarithm does not change the location of the maximum, and it simplifies calculations: ℓ(θ)=log(L(θ))

**Optimization**:

* Once the log-likelihood function is defined, optimization techniques (such as gradient descent or more sophisticated methods) are used to find the values of the parameters that maximize the log-likelihood.

**Where is it applied best?**

```r
#R implementation
# Simulate some data from a geometric distribution
set.seed(123)  # for reproducibility
data <- dgeom(100, prob = 0.3)

# Likelihood function for a geometric distribution
likelihood_function <- function(p, x) {
  prob = p^(x - 1) * (1 - p)
  return(prod(prob))
}

# Log-likelihood function
log_likelihood_function <- function(p, x) {
  log_likelihood = sum((x - 1) * log(p) + log(1 - p))
  return(log_likelihood)
}

# Maximize the log-likelihood to find the MLE
mle_result <- optim(par = 0.5,
                    fn = log_likelihood_function,
                    x = data, method = "Brent",
                    lower = 0, upper = 1)

# Display the MLE estimate
cat("Maximum Likelihood Estimate (MLE) for p:", mle_result$par, "\n")
```
