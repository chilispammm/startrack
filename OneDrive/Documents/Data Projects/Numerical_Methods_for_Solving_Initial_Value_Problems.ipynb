## 1. Euler Method
- This is the simplest method, which takes the slope at the current point to approximate the next point. It tends to be less accurate but easy to implement.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function f(t, y)
def f(t, y):
    return t - y

# Euler method implementation
def euler_method(t0, y0, h, steps):
    t_values = [t0]
    y_values = [y0]

    for _ in range(steps):
        y_next = y_values[-1] + h * f(t_values[-1], y_values[-1])
        t_next = t_values[-1] + h

        y_values.append(y_next)
        t_values.append(t_next)

    return t_values, y_values

# Initial values
t0, y0, h, steps = 0, 1, 0.1, 20

# Get the Euler solution
t_euler, y_euler = euler_method(t0, y0, h, steps)
t_euler, y_euler
```

## 2. Heun Method
- This one's an improvement over the Euler method, as it averages the slope at the beginning and the predicted end of the interval, giving a better approximation.

```python
# Heun method implementation
def heun_method(t0, y0, h, steps):
    t_values = [t0]
    y_values = [y0]

    for _ in range(steps):
        y_pred = y_values[-1] + h * f(t_values[-1], y_values[-1])
        y_next = y_values[-1] + (h / 2) * (f(t_values[-1], y_values[-1]) + f(t_values[-1] + h, y_pred))

        t_next = t_values[-1] + h
        y_values.append(y_next)
        t_values.append(t_next)

    return t_values, y_values

# Get the Heun solution
t_heun, y_heun = heun_method(t0, y0, h, steps)
t_heun, y_heun
```

## 3. Runge-Kutta (4th Order) Method
- This is a highly accurate method for solving IVPs. It uses weighted averages of slopes at several points within the interval to get a more accurate result.

```python
# Runge-Kutta 4th Order method implementation
def runge_kutta_method(t0, y0, h, steps):
    t_values = [t0]
    y_values = [y0]

    for _ in range(steps):
        k1 = f(t_values[-1], y_values[-1])
        k2 = f(t_values[-1] + h / 2, y_values[-1] + h / 2 * k1)
        k3 = f(t_values[-1] + h / 2, y_values[-1] + h / 2 * k2)
        k4 = f(t_values[-1] + h, y_values[-1] + h * k3)

        y_next = y_values[-1] + (h / 6) * (k1 + 2 * k2 + 2 * k3 + k4)
        t_next = t_values[-1] + h

        y_values.append(y_next)
        t_values.append(t_next)

    return t_values, y_values

# Get the Runge-Kutta solution
t_rk, y_rk = runge_kutta_method(t0, y0, h, steps)
t_rk, y_rk
```

## 4. Exact Solution and Plotting

```python
# Exact solution
def exact_solution(t):
    return t - 1 + 2 * np.exp(-t)

# Generate t values for exact solution
t_exact = np.linspace(0, 2, 100)
y_exact = exact_solution(t_exact)

# Plotting the solutions
plt.figure(figsize=(10, 6))
plt.plot(t_exact, y_exact, label="Exact Solution", color="black", linestyle="--")
plt.plot(t_euler, y_euler, label="Euler Method", marker='o')
plt.plot(t_heun, y_heun, label="Heun Method", marker='x')
plt.plot(t_rk, y_rk, label="Runge-Kutta 4th Order", marker='s')
plt.xlabel("t")
plt.ylabel("y")
plt.title("Numerical Methods for Solving IVP")
plt.legend()
plt.grid()
plt.show()
```

### Key Insights:



The plot compares the performance of 3 numerical methods (Euler, Heun, and Runge-Kutta 4th order) against the exact solution of an initial value problem (IVP). The time interval considered is from 0 to 2.

**Individual Method Performance:**

1. **Euler Method:**
   - It shows significant deviation from the exact solution, especially at the beginning and end of the time interval.
   - The error accumulates over time, leading to a less accurate approximation.

2. **Heun Method:**
   - This one provides a more accurate approximation than the Euler method.
   - The error is reduced compared to Euler, but it still deviates from the exact solution, particularly towards the end.

3. **Runge-Kutta 4th Order:**
   - This one offers the highest accuracy among the three methods.
   - The approximation closely follows the exact solution throughout the entire time interval.

**Error Analysis:**

- The Euler method is known for its simplicity but suffers from low accuracy due to its large truncation error.
- The Heun method improves accuracy by using a predictor-corrector approach, reducing the truncation error.
- The Runge-Kutta 4th order method is highly accurate due to its sophisticated algorithm that minimizes truncation error.



**Conclusions:**

- For solving IVPs, the **Runge-Kutta 4th order method** generally provides the most accurate results.
- The choice of method depends on the desired level of accuracy and computational cost. If high accuracy is critical, Runge-Kutta 4th order is preferred. If computational efficiency is a priority, Euler or Heun methods might be considered, but with a trade-off in accuracy.
