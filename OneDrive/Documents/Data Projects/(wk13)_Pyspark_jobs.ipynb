```python
!pip install understatapi --quiet
import pandas as pd
import understatapi

client = understatapi.UnderstatClient()
```

```python
# Data from EPL to retrieve a list of matches
league_data = client.league(league="EPL").get_match_data(season="2024")
league_data[0]
#df_league = pd.DataFrame(league_data)
#df_league.head()
```

```python
import pandas as pd

# Assuming league_data is already populated as in your previous code
# league_data = client.player("Manchester United").get_match_data(season="2024")

# Create a list of dictionaries with the desired columns
extracted_data = []
for match in league_data:
    extracted_data.append({
        'ID': match['id'],
        'h': match['h'],
        'a': match['a'],
        'datetime': match['datetime'],
        'goals': match['goals'],
        'xG': match['xG']
    })

# Create a pandas DataFrame from the extracted data
df_matches = pd.DataFrame(extracted_data)

# Display the DataFrame
df_matches.head()
```

```python
# Extract data from the 'h' column
h_data = df_matches['h'].apply(pd.Series)
h_data = h_data.add_prefix('h_')  # Add prefix to avoid column name conflicts

# Extract data from the 'a' column
a_data = df_matches['a'].apply(pd.Series)
a_data = a_data.add_prefix('a_')  # Add prefix to avoid column name conflicts

# Extract data from the 'goals' column
goals_data = df_matches['goals'].apply(pd.Series)
goals_data = goals_data.add_prefix('goals_')  # Add prefix

# Extract data from the 'xg' column
xg_data = df_matches['xG'].apply(pd.Series)
xg_data = xg_data.add_prefix('xG_')  # Add prefix


# Concatenate the extracted data with the original DataFrame
df_expanded = pd.concat([df_matches, h_data, a_data], axis=1)
# Concatenate with the existing df_expanded
df_expanded = pd.concat([df_expanded, goals_data, xg_data], axis=1)

# Drop the original 'h' and 'a' columns
df_expanded = df_expanded.drop(columns=['h', 'a'])

# Drop the original 'goals' and 'xg' columns
df_expanded = df_expanded.drop(columns=['goals', 'xG'])

df_expanded.head()
```

```python
# Select the desired columns
team_data = df_expanded[['ID', 'datetime', 'h_title', 'a_title', 'goals_h', 'goals_a', 'xG_h', 'xG_a']]
team_data.head()
```

#### Understanding Expected Goals (xG)
##### What is xG?
* xG measures the quality of scoring chances by assigning each shot a probability of becoming a goal. Higher xG = better chances.

##### Why it matters:

* More reliable than actual goals for predicting future performance

* Shows which teams create the best chances (not just lucky scorers)

### 2. Low-Level: RDD Operations


```python
# Create a PySpark DataFrame from the pandas DataFrame 'team_data'
football_df = spark.createDataFrame(team_data)

# Cast xG columns to float type
football_df = football_df.withColumn("xG_h", football_df["xG_h"].cast("float"))
football_df = football_df.withColumn("xG_a", football_df["xG_a"].cast("float"))

# Show the first few rows
football_df.show(5)

# Convert DataFrame to RDD
football_rdd = football_df.rdd

# Count total matches
total_matches = football_rdd.count()
print(f"Total matches: {total_matches}")

# Get matches with high xG difference (exciting matches)
exciting_matches = football_rdd.filter(
    lambda row: abs(row["xG_h"] - row["xG_a"]) > 2
).map(
    lambda row: (row["h_title"], row["a_title"], row["xG_h"], row["xG_a"])
)

print("\nExciting matches (xG difference > 2):")
for match in exciting_matches.take(5):
    print(f"{match[0]} vs {match[1]}: {match[2]:.2f}-{match[3]:.2f} xG")
```

### 3. Mid-Level: DataFrame and SQL Operations
#### 3.1) Team Performance Analysis

```python
from pyspark.sql.functions import col, sum, count, when, lit

# Create home and away views
home_stats = football_df.groupBy("h_title").agg(
    sum("goals_h").alias("home_goals_scored"),
    sum("goals_a").alias("home_goals_conceded"),
    sum("xG_h").alias("home_xG"),
    sum("xG_a").alias("home_xGA"),
    count(lit(1)).alias("home_matches")
)

away_stats = football_df.groupBy("a_title").agg(
    sum("goals_a").alias("away_goals_scored"),
    sum("goals_h").alias("away_goals_conceded"),
    sum("xG_a").alias("away_xG"),
    sum("xG_h").alias("away_xGA"),
    count(lit(1)).alias("away_matches")
)

# Join and calculate totals
team_stats = home_stats.join(away_stats, home_stats.h_title == away_stats.a_title) \
    .select(
        col("h_title").alias("team"),
        (col("home_goals_scored") + col("away_goals_scored")).alias("total_goals_scored"),
        (col("home_goals_conceded") + col("away_goals_conceded")).alias("total_goals_conceded"),
        (col("home_xG") + col("away_xG")).alias("total_xG"),
        (col("home_xGA") + col("away_xGA")).alias("total_xGA"),
        (col("home_matches") + col("away_matches")).alias("total_matches")
    )

# Show teams sorted by xG difference (attack strength)
team_stats.withColumn("xG_difference", col("total_xG") - col("total_xGA")) \
    .orderBy("xG_difference", ascending=False) \
    .show()
```

* Bournemouth (+17.6) - Surprisingly effective attack

#### 3.2) SQL Query for Match Outcomes


```python
# Register DataFrame as a temporary view
football_df.createOrReplaceTempView("matches")

# SQL query to find matches where the result didn't match xG expectation
surprising_results = spark.sql("""
    SELECT
        datetime,
        h_title as home_team,
        a_title as away_team,
        goals_h as home_goals,
        goals_a as away_goals,
        xG_h as home_xG,
        xG_a as away_xG,
        CASE
            WHEN goals_h > goals_a AND xG_h < xG_a THEN 'Home overperformance'
            WHEN goals_h < goals_a AND xG_h > xG_a THEN 'Away overperformance'
            ELSE 'Expected result'
        END as result_type
    FROM matches
    WHERE (goals_h > goals_a AND xG_h < xG_a) OR (goals_h < goals_a AND xG_h > xG_a)
    ORDER BY ABS(xG_h - xG_a) DESC
""")

surprising_results.show(10)
```

### 4. High-Level: Advanced Analytics
#### 4.1 Expected vs Actual Performance

```python
from pyspark.sql.functions import avg

# Calculate performance metrics
performance = football_df.withColumn("home_performance", col("goals_h") - col("xG_h")) \
    .withColumn("away_performance", col("goals_a") - col("xG_a"))

# Average performance by team (home and away)
home_perf = performance.groupBy("h_title") \
    .agg(avg("home_performance").alias("avg_home_performance"))

away_perf = performance.groupBy("a_title") \
    .agg(avg("away_performance").alias("avg_away_performance"))

# Combine results
team_performance = home_perf.join(away_perf, home_perf.h_title == away_perf.a_title) \
    .select(
        col("h_title").alias("team"),
        col("avg_home_performance"),
        col("avg_away_performance"),
        ((col("avg_home_performance") + col("avg_away_performance"))/2).alias("avg_performance")
    ) \
    .orderBy("avg_performance", ascending=False)

team_performance.show()
```

#### 4.2 Match Predictions Using PySpark and Understat xG Data
Here we build a basic match prediction model using your Understat data with PySpark. The expected goals (xG) metric is particularly valuable for this purpose as it provides a more stable performance indicator than actual goals.


We'll use these key techniques:

* **Feature Engineering**: Create predictive features from historical xG data

* **Model Training**: Build a classifier to predict match outcomes

* **Evaluation**: Assess model performance

##### 4.2.1 Feature Engineering

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

# Calculate rolling averages for team performance
window_spec = Window.partitionBy("team").orderBy("date").rowsBetween(-5, -1)

# Create home and away team views
home_team_stats = football_df.select(
    F.col("h_title").alias("team"),
    F.col("datetime").alias("date"),
    F.col("xG_h").alias("xG_for"),
    F.col("xG_a").alias("xG_against"),
    F.col("goals_h").alias("goals_for"),
    F.col("goals_a").alias("goals_against"),
    F.lit("home").alias("venue")
)

away_team_stats = football_df.select(
    F.col("a_title").alias("team"),
    F.col("datetime").alias("date"),
    F.col("xG_a").alias("xG_for"),
    F.col("xG_h").alias("xG_against"),
    F.col("goals_a").alias("goals_for"),
    F.col("goals_h").alias("goals_against"),
    F.lit("away").alias("venue")
)

# Combine home and away stats
team_stats = home_team_stats.union(away_team_stats)

# Calculate rolling averages
team_features = team_stats.withColumn(
    "avg_xG_for", F.avg("xG_for").over(window_spec)
).withColumn(
    "avg_xG_against", F.avg("xG_against").over(window_spec)
).withColumn(
    "avg_goals_for", F.avg("goals_for").over(window_spec)
).withColumn(
    "avg_goals_against", F.avg("goals_against").over(window_spec)
)

# Join features back to original matches
matches_with_features = football_df.alias("matches").join(
    team_features.alias("home").select(
        "team", "date",
        "avg_xG_for", "avg_xG_against",
        "avg_goals_for", "avg_goals_against"
    ),
    (F.col("matches.h_title") == F.col("home.team")) &
    (F.col("matches.datetime") == F.col("home.date")),
    "left"
).join(
    team_features.alias("away").select(
        "team", "date",
        "avg_xG_for", "avg_xG_against",
        "avg_goals_for", "avg_goals_against"
    ),
    (F.col("matches.a_title") == F.col("away.team")) &
    (F.col("matches.datetime") == F.col("away.date")),
    "left"
)

# Define target variable (1 if home wins, 0 otherwise)
matches_with_target = matches_with_features.withColumn(
    "home_win",
    F.when(F.col("goals_h") > F.col("goals_a"), 1).otherwise(0)
)

# Show the enhanced dataset
matches_with_target.select(
    "h_title", "a_title", "goals_h", "goals_a", "home_win",
    "home.avg_xG_for", "home.avg_xG_against",
    "away.avg_xG_for", "away.avg_xG_against"
).show(5)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

# Calculate rolling averages with proper aliasing
window_spec = Window.partitionBy("team").orderBy("date").rowsBetween(-5, -1)

# Create home team features with clear aliases
home_features = team_features.filter(F.col("venue") == "home") \
    .select(
        F.col("team").alias("home_team"),
        F.col("date").alias("match_date"),
        F.col("avg_xG_for").alias("home_avg_xG_for"),
        F.col("avg_xG_against").alias("home_avg_xG_against"),
        F.col("avg_goals_for").alias("home_avg_goals_for"),
        F.col("avg_goals_against").alias("home_avg_goals_against")
    )

# Create away team features with clear aliases
away_features = team_features.filter(F.col("venue") == "away") \
    .select(
        F.col("team").alias("away_team"),
        F.col("date").alias("match_date"),
        F.col("avg_xG_for").alias("away_avg_xG_for"),
        F.col("avg_xG_against").alias("away_avg_xG_against"),
        F.col("avg_goals_for").alias("away_avg_goals_for"),
        F.col("avg_goals_against").alias("away_avg_goals_against")
    )

# Join with original matches using proper column references
matches_with_features = football_df.alias("matches") \
    .join(
        home_features.alias("home"),
        (F.col("matches.h_title") == F.col("home.home_team")) &
        (F.col("matches.datetime") == F.col("home.match_date")),
        "left"
    ) \
    .join(
        away_features.alias("away"),
        (F.col("matches.a_title") == F.col("away.away_team")) &
        (F.col("matches.datetime") == F.col("away.match_date")),
        "left"
    )

# Define target variable (1 if home wins, 0 otherwise)
matches_with_target = matches_with_features.withColumn(
    "home_win",
    F.when(F.col("goals_h") > F.col("goals_a"), 1).otherwise(0)
)

# Show the schema to verify column names
matches_with_target.printSchema()
```

##### 4.2.1 Model Training
We'll use PySpark MLlib to train a classifier:

```python
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.classification import LogisticRegression
from pyspark.ml import Pipeline

from pyspark.ml.feature import VectorAssembler, Imputer


# Create feature vector, but handle null values with Imputer
# Correct feature columns - use the aliases we created
feature_cols = [
    "home_avg_xG_for", "home_avg_xG_against",
    "away_avg_xG_for", "away_avg_xG_against"
]

# Create an Imputer to replace nulls with the mean
imputer = Imputer(
    inputCols=feature_cols,
    outputCols=[f"{c}_imputed" for c in feature_cols],
    strategy="mean"  # Replace with mean, you can use other strategies too
)

# Update feature columns to use imputed values
feature_cols_imputed = [f"{c}_imputed" for c in feature_cols]

# Create feature vector using imputed columns
assembler = VectorAssembler(
    inputCols=feature_cols_imputed,
    outputCol="features"
)


# Split data
train, test = matches_with_target.randomSplit([0.8, 0.2], seed=42)

# Impute missing values in train and test data
train = imputer.fit(train).transform(train)
test = imputer.fit(test).transform(test)

# Logistic regression model
lr = LogisticRegression(
    featuresCol="features",
    labelCol="home_win",
    predictionCol="prediction",
    probabilityCol="probability"
)

# Pipeline
pipeline = Pipeline(stages=[assembler, lr])

# Train model
model = pipeline.fit(train)

# Make predictions
predictions = model.transform(test)
predictions.select("h_title", "a_title", "probability", "prediction").show(5)
```
